# Train Task 3 Transformer

This notebook trains the decoder-only Transformer on token sequences.

```python
import pickle
from pathlib import Path

token_path = Path('data/tokenized/tokens_train.pkl')
with open(token_path, 'rb') as f:
    payload = pickle.load(f)
vocab_size = max(max(seq) for seq in payload['sequences']) + 1
pad_token_id = payload.get('pad_token_id', 0)
print('vocab_size:', vocab_size)
print('pad_token_id:', pad_token_id)
```
